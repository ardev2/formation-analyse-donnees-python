## Partie 1 : créer l'environnement virtuel

Les postes sont sous Ubuntu, avec **Python 3.12 déjà installé** : on n'installe pas Python. On crée seulement un **environnement virtuel**, un dossier `.venv` qui contiendra les bibliothèques de ce TP. Le cours y reviendra en détail juste après.

Dans un terminal :

1. Vérifie la version de Python :
   ```bash
   python3 --version
   ```
   Résultat attendu : `Python 3.12.x`.
2. Place-toi dans le dossier de ce TP, puis crée l'environnement virtuel :
   ```bash
   cd chemin/vers/exercices/tp00-verification-environnement
   sudo apt install python3.12-venv
   python3 -m venv .venv
   ```
3. Active-le :
   ```bash
   source .venv/bin/activate
   ```
   L'invite de commande commence maintenant par `(.venv)`.
4. Installe JupyterLab dans l'environnement, puis lance-le :
   ```bash
   python -m pip install jupyterlab
   jupyter lab
   ```
5. Dans JupyterLab, ouvre `tp00.ipynb`.
